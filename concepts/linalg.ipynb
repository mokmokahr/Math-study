## Orthogonal Projection


$$\operatorname{proj}_{\mathbf u}(\mathbf v)
=\frac{\mathbf v\cdot\mathbf u}{\mathbf u\cdot\mathbf u}\mathbf u$$

The projection of v onto u is a vector parallel to u. Its length is |u · v| / ||u||.
